```python
sum = 0

for num in range(101):
    if num % 2 == 0:
        sum += num
print(f' The sum of even numbers is {sum}')
```

```python
# Assignment 3
# Sum all the odd numbers between 1 and 100 using each of the two versions of while loop
# Version 1: Increment by 2
total = 0
num = 1

while num <= 100:
    total += num
    num += 2

print("Sum of odd numbers :", total)
```

```python
# Version 2: Check odd using %
total = 0
num = 1

while num <= 100:
    if num % 2 != 0:
        total += num
    num += 1

print("Sum of odd numbers :", total)
```

```python
# Version 3: Using a list of odd numbers
odd_numbers = list(range(1, 101, 2))
i = 0
total = 0

while i < len(odd_numbers):
    total += odd_numbers[i]
    i += 1

print("Sum of odd numbers :", total)
```

```python
# Univariate Analysis
# 1. Trend of Temperature over time
# Load Libraries
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Ensure datetime is properly parsed
nyc_Dataset_df['EST'] = pd.to_datetime(nyc_Dataset_df['EST'], errors='coerce')

# Time Series Analysis
# Sort by datetime
nyc_Dataset_df = nyc_Dataset_df.sort_values("EST")

# Temperature trend
fig = px.line(nyc_Dataset_df, x="EST", y="Temperature", title="Temperature Over Time")
fig.show()

# Rolling Average (30 days)
nyc_Dataset_df['Temp_Rolling'] = nyc_Dataset_df['Temperature'].rolling(30).mean()
fig = px.line(nyc_Dataset_df, x="EST", y="Temp_Rolling", title="30-Day Rolling Average of Temperature")
fig.show()

# Monthly Average Temperature
monthly_avg = nyc_Dataset_df.resample("M", on="EST")["Temperature"].mean().reset_index()
fig = px.line(monthly_avg, x="EST", y="Temperature", markers=True, title="Monthly Average Temperature")
fig.show()
```
